# Файлы

Мы уже многое умеем делать с данными. Но на практике это оказывается совершенно бесполезно, если мы не умеем их хранить. Для хранения данных используются разнообразные файлы: текстовые, табличные и другие.

## Чтение файлов

Чтение файлов происходит при помощи функции `open()`, в аргумент которой нужно передать путь к файлу (если файл размещен в той же папке, что наша программа, то просто название файла).

````python
with open("text.txt", encoding="utf-8") as f:
    text = f.read()   # по-умолчанию .read() читает файл целиком
````

Смотрим на **синтаксис**: файл в аргументах `open()` мы открываем "как f", то есть результат выполнения функции - содержимое файла - сразу же запишется в переменную f в виде одной сплошной строки.

*Эта переменная, как и любая другая, может называться как угодно. В случае с чтением файлов мы содержимое этой переменной будем использовать сразу же и только единожды, поэтому на названии можно сэкономить и не называть переменную по смыслу.*

Оператором with мы обозначаем, что дальнейший *вложенный* блок будет выполнен с теми данными, которые мы записали в f.

А поскольку часть кода будет *вложенной* - двоеточие и отступы тут как тут.

Данные из файла записали в f, но одного этого действия недостаточно. Нужно еще обязательно дешифровать данные при помощи метода `.read()`. После этой строчки переменной, в которую мы записали текст, можно пользоваться, как любой другой. Ничто не изменится и если мы вынесем дальнейшие манипуляции с ней за пределы вложенного кода.

````python
print(text)
````

## Обработка данных файла

Если мы работаем с текстовым файлом, часто нам будет нужен не сплошной текст из него, а список либо слов, либо строк файла. Со словами все понятно: метод `.split()` без аргумента разделит текст по пробелам.

````python
print(text.split())
````

Для деления на строки будем использовать тот же `.split()`, но в аргументы придется передать специальный символ переноса строки: `\n`.

````python
print(text.split('\n'))
````

Таблицы, с которыми не нужно работать вручную (то есть читать их глазами), часто сохраняют в форматах `CSV` (значения, разделенные запятыми, расширение `.csv`) или `TSV` (значения, разделенные табуляциями, расширение `.tsv`). Для обработки таких таблиц файл читают точно так же, как текстовый, затем разбивают на строки при помощи `.split('\n')` и потом каждую строку - на ячейки при помощи `.split('\t')` (`\t` - специальный символ табуляции).

## Пути к файлам

У нас есть файловая система. На диске `C` есть папка `Users`, в которой хранятся папки `Anna` и `Yuri`, а также файл `common_file.txt`. В папке `Anna` есть папка `examples` с единственным файлом `example.txt` и еще два файла: `fix_grammar.py` и `diploma.txt`. В папке `Yuri` хранится только один файл `my_diploma.txt`.

```
C:/
    Users/
        Anna/
            examples/
                example.txt
            fix_grammar.py
            diploma.txt
        Yuri/
            my_diploma.txt
        common_file.txt
```

Пути к файлам бывают абсолютные или относительные.

**Абсолютные пути** - те, которые начинаются с корня файловой системы. Например, в Windows это может быть диск C: `C:/Users` переходит к папке Users на диске C. *На MacOS абсолютный путь будет начинаться просто со слеша, без названия диска: `/Users` перейдет к такой же папке Users в корне сиситемы.* Для перехода по относительному пути к файлу `fix_grammar.py` нам нужно будет указать такой путь: `C:/Users/Anna/fix_grammar.py`.

**Относительные пути** определяются относительно расположения файла, из которого запускается программа. Для всех примеров ниже считаем, что мы находимся в папке `Anna`.
* Если нужный файл лежит в той же папке, что и наша программа, мы напишем просто название этого файла. Для перехода к `fix_grammar.py` мы так и укажем: `fix_grammar.py`.
* Если в папке, где мы находимся, лежит другая папка, а в ней - нужный файл, напишем название папки и через слеш - название файла. Для перехода к файлу `example.txt` укажем `examples/example.txt`.
* Если для того, чтобы попасть к файлу нам нужно выйти из текущей папки на уровень выше, используем оператор `..`: `../file.txt`. Для перехода к файлу `common_file.txt` напишем `../common_file.txt`.
