## Библиотеки

Готовый инструмент, доступный для использования, называют `модулем`. А коллекцию модулей, относящихся к смежным задачам, называют `библиотекой`.

Часть библиотек, наиболее часто используемые, устанавливаются автоматически при установке Python. Например, библиотека `string` для работы со строками, `math` содержит разнообразные математические операции, `datetime` позволяет работать с датой и временем, `json` - с данными формата JSON (формат хранит списки словарей), `os`, `sys` и `pathlib` позволяют удобнее работать с обращениями кода к системе компьютера и файловой структурой.

Посмотрим, как это работает, на примере библиотеки `string`.

````python
# сначала библиотеку нужно импортировать
import string

# а затем просто использовать функции, хранящиеся в этой библиотеке
print(string.punctuation)
````

Функция `punctuation` выдала нам строку из всех возможных символов пунктуации. Обратите внимание на то, как мы ее вызвали: сначала мы указали источник, в котором функция хранится - библиотеку `string`, а затем через точку написали название функции. Причем скобки у функции мы не писали.

Доступный в библиотеке функционал, особенности работы и синтаксис всегда описаны в документации библиотеки. Для того, чтобы ее найти, достаточно выполнить поисковый запрос в интернете.

Посмотрим на другую библиотеку:

````python
import math

print(math.factorial(5))
````

Многие полезные библиотеки не устанавливаются вместе с Python. В таком случае их нужно скачать. Для этого используется установщик `pip`. Почитать разное о нем можно [здесь](https://pypi.org/project/pip/). В консоли пищется `pip install <название-библиотеки>`.

Основная сущность, с котрой работает библиотека - это датафрейм (`DataFrame`). На человеческом - таблица.

Датафрейм можно получить различными способами, но основных два:

* создать из словаря, списка или чего-то похожего

````python
dct = {
    'age': [12, 5, 15, 25],
    'height': [140, 110, 165, 180]
}
pd.DataFrame(dct)
````

````python
lst = [
    [12, 140],
    [5, 110],
    [15, 165],
    [25, 180]
]
pd.DataFrame(lst, columns=['age', 'height'])
````

* загрузить из файла (csv, tsv, excel etc.)
  * CSV. Используется функция read_csv(). Аргумент file является строкой, в которой записан путь до файла с дата-сетом. Для записи данных из DataFrame в CSV-файл используется метод to_csv(file).
  * Excel. Используется функция read_excel(). Для записи данных из DataFrame в Excel-файл используется метод to_excel().
  * JSON. Используется функция read_json(). Для записи данных из DataFrame в JSON используется метод to_json().
  * Онлайн-таблицы (на страницах сайта). Используется функция read_html().

Код для чтения таблицы, сохраненной в файле `table.csv`, будет:

```python
df = pd.read_csv('table.csv')
```

После этого датафрейм из таблицы будет сохранен в переменную df, с которой мы сможем дальше работать при помощи методов библиотеки.

````python
df.head(2) # покажет две верхние строки таблицы
df.sample(2) # покажет две случайные строки таблицы
df.tail(2) # покажет две нижние строки таблицы
````

# Еще возможности

`df.columns` выведет список названий столбцов датафрейма.

А написав в квадратных скобках название столбца, можно получить его данные: ‍`df['Reviewer']`.

Похожим образом, в квадратных скобках через `df.loc`, можно указывать индексы строк и даже сразу и индексы строк, и названия столбцов - и использовать при этом срезы: `df.loc[5:10, 'Rating':'Metadata']`.

А такой код\
`df.loc[df.Restaurant == 'Beyond Flavours', ['Restaurant', 'Rating', 'Metadata']].head(3)` \
покажет нам данные из трех столбцов, предварительно отфильтровав таблицу: оставив только те строки, в которых в столбце "Restaurant" будет указано "Beyond Flavours".

Можно даже использовать несколько условий фильтрации в одной строке. Для связи отдельных условий могут использоваться `&` (и), `|` (или), `^` (либо) или `~` (не), а каждое условие нужно заключить в круглые скобки.

`df.drop` удаляет столбцы (изучите синтаксис подробнее в документации), а добавить столбцы можно примерно так же, как мы добавляем новую пару ключ-значение в словарь: пишем переменную, в которой находится датафрейм, в квадратных скобках название столбца, а затем через "равно" указываем данные, которые нужно записать в столбец.

`df.value_counts('Restaurant')` сделает таблицу со статистикой: перечислит в одном столбце все возможные значения столбца Restaurant и рядом укажет, сколько раз каждое значение встретилось.

`df['Restaurant'].unique()` вернет список уникальных значений, содержащихся в столбце Restaurant.

# Запись файлов

Мы уже умеем читать файлы. Запись делается примерно так же, с небольшим отличием: среди аргументов функции `open` должен быть флаг, который сообщит питону, что файл мы хотим не читать, а записывать. От того, какой именно флаг, зависит, как будет происходить запись: `w` после создания или открытия существующего файла его полностью перезапишет, а `a` будет дозаписывать переданные данные после уже имеющихся. Например:

````python
with open ('my_file.txt', 'w') as f:
    f.write("Hello, World!")
````
