```python
with open('people.txt', encoding='utf-8') as f:
    content = f.read()

lines = content.split('\n')
new_lines = [line.split('\t') for line in lines if line]

columns = [k for k in new_lines[0]]
data_list = []

for line in new_lines[1:]:
    i = 0
    line_dict = {}
    while i < len(columns):
        if i < len(line):
            line_dict[columns[i]] = line[i]
        i += 1
    data_list.append(line_dict)
    
print(data_list)
```
